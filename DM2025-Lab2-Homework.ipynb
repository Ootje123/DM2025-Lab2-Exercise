# **Competition Code**

## 1. Preprocessing Steps

```python
import pandas as pd
import json
import re
# === Load raw Phase 3 data ===
raw = pd.read_json("./data/final_posts.json")

# Flatten JSON structure
def extract_post(row):
    try:
        src = row["root"]["_source"]["post"]
        return pd.Series({
            "id": src["post_id"],
            "text": src["text"]
        })
    except:
        return pd.Series({"id": None, "text": None})

posts = raw.apply(extract_post, axis=1)

# Load other files
ids   = pd.read_csv("./data/data_identification.csv")      
labels = pd.read_csv("./data/emotion.csv")                 

# === Split train/test ids ===
train_ids = ids[ids["split"] == "train"]
test_ids  = ids[ids["split"] == "test"]

# === Build train and test dataframes ===
train = (
    train_ids
    .merge(posts, on="id", how="left")
    .merge(labels, on="id", how="left")
)

test = (
    test_ids
    .merge(posts, on="id", how="left")
)

print("Train shape:", train.shape)
print("Test shape:", test.shape)
train.head()

def clean(t):
    if pd.isna(t):
        return ""
    t = t.lower()
    t = re.sub(r"http\S+", " ", t)      # remove URLs
    t = re.sub(r"[^a-zA-Z\s]", " ", t)  # keep letters
    t = re.sub(r"\s+", " ", t).strip()  # normalize spaces
    return t

train["clean_text"] = train["text"].apply(clean)
test["clean_text"]  = test["text"].apply(clean)
```

## 2. Feature Engineering Steps

```python
from sklearn.feature_extraction.text import TfidfVectorizer

vectorizer = TfidfVectorizer(
    max_features=70000,
    min_df=2,
    ngram_range=(1, 2),
    sublinear_tf=True
)

X_train = vectorizer.fit_transform(train["clean_text"])
y_train = train["emotion"]

print("Train TF-IDF shape:", X_train.shape)
```

```python
# Ensure test rows are in the same order as samplesubmission
sample = pd.read_csv("./data/samplesubmission.csv")   # id,emotion (dummy)
test_ordered = sample[["id"]].merge(
    test[["id", "clean_text"]],
    on="id",
    how="left"
)

X_test = vectorizer.transform(test_ordered["clean_text"])
X_test.shape
```

## 3. Model Implementation Steps

```python
from sklearn.svm import LinearSVC
from sklearn.metrics import classification_report, f1_score

clf = LinearSVC(
    C=1.5,
    class_weight="balanced"
)

clf.fit(X_train, y_train)

# Optional check on training
train_pred = clf.predict(X_train)
print("Train macro-F1:", f1_score(y_train, train_pred, average="macro"))

# Predict on training data (proxy for performance)
train_pred = clf.predict(X_train)

# 1. Full classification report table
report = classification_report(
    y_train,
    train_pred,
    output_dict=True
)

results_df = pd.DataFrame(report).transpose()

# 2. Print clean table
print("===== RESULTS TABLE (Training Evaluation) =====")
display(results_df)

# 3. Print the macro F1 clearly
macro_f1 = f1_score(y_train, train_pred, average="macro")
print("\nMacro F1 Score:", round(macro_f1, 4))
```

```python
# Ensure correct order
sample = pd.read_csv("./data/samplesubmission.csv")

test_ordered = sample[["id"]].merge(
    test[["id", "clean_text"]],
    on="id",
    how="left"
)

X_test = vectorizer.transform(test_ordered["clean_text"])
y_pred = clf.predict(X_test)

submission = sample.copy()
submission["emotion"] = y_pred
submission.to_csv("submission.csv", index=False)

print("Saved submission.csv ✔")
submission.head()
```
