# **Competition Code**

## 1. Preprocessing Steps

```python
import pandas as pd
import json
import re
# === Load raw Phase 3 data ===
raw = pd.read_json("./data/final_posts.json")

# Flatten JSON structure
def extract_post(row):
    try:
        src = row["root"]["_source"]["post"]
        return pd.Series({
            "id": src["post_id"],
            "text": src["text"]
        })
    except:
        return pd.Series({"id": None, "text": None})

posts = raw.apply(extract_post, axis=1)

# Load other files
ids   = pd.read_csv("./data/data_identification.csv")      
labels = pd.read_csv("./data/emotion.csv")                 

# === Split train/test ids ===
train_ids = ids[ids["split"] == "train"]
test_ids  = ids[ids["split"] == "test"]

# === Build train and test dataframes ===
train = (
    train_ids
    .merge(posts, on="id", how="left")
    .merge(labels, on="id", how="left")
)

test = (
    test_ids
    .merge(posts, on="id", how="left")
)

print("Train shape:", train.shape)
print("Test shape:", test.shape)
train.head()

def clean(t):
    if pd.isna(t):
        return ""
    t = t.lower()
    t = re.sub(r"http\S+", " ", t)      # remove URLs
    t = re.sub(r"[^a-zA-Z\s]", " ", t)  # keep letters
    t = re.sub(r"\s+", " ", t).strip()  # normalize spaces
    return t

train["clean_text"] = train["text"].apply(clean)
test["clean_text"]  = test["text"].apply(clean)
```

## 2. Feature Engineering Steps

```python
from sklearn.feature_extraction.text import TfidfVectorizer, ENGLISH_STOP_WORDS
from scipy.sparse import hstack

# === Word-level TF–IDF ===
vectorizer = TfidfVectorizer(
    max_features=40000,
    min_df=3,
    ngram_range=(1, 2),
    sublinear_tf=True,
)

X_train_word = vectorizer.fit_transform(train["clean_text"])
y_train = train["emotion"]

# === Character-level TF–IDF ===
char_vectorizer = TfidfVectorizer(
    analyzer='char',
    ngram_range=(3, 6),
    min_df=3,
    max_features=20000,
)

X_train_char = char_vectorizer.fit_transform(train["clean_text"])

# === Combine safely (sparse!) ===
X_train = hstack([X_train_word, X_train_char])
print("Train shape:", X_train.shape)
```

```python
X_train = hstack([X_train_word, X_train_char])
print("Train shape:", X_train.shape)

# === Make test in the same order ===
sample = pd.read_csv("./data/samplesubmission.csv")

test_ordered = sample[["id"]].merge(
    test[["id", "clean_text"]],
    on="id",
    how="left"
)

# === Transform BOTH word + char!!! ===
X_test_word = vectorizer.transform(test_ordered["clean_text"])
X_test_char = char_vectorizer.transform(test_ordered["clean_text"])

X_test = hstack([X_test_word, X_test_char])
print("Test shape:", X_test.shape)
```

## 3. Model Implementation Steps

```python
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report, f1_score
from sklearn.linear_model import LogisticRegression

# 3. Train/validation split
X_tr, X_val, y_tr, y_val = train_test_split(
    X_train, y_train,
    test_size=0.2,
    random_state=42,
    stratify=y_train
)

# 4. Hyperparameter tuning for C
best_c = None
best_score = 0

for c in [2,2.25, 2.5]:
    clf = LogisticRegression(
        max_iter=2000,
        C=c,
        multi_class="multinomial"
    )
    clf.fit(X_tr, y_tr)
    val_pred = clf.predict(X_val)
    score = f1_score(y_val, val_pred, average="macro")
    print(f"C={c}: F1={score:.4f}")
    
    if score > best_score:
        best_score = score
        best_c = c

print("Best C value:", best_c)

# 5. Refit with best C on the same train split (for evaluation)
clf = LogisticRegression(
    max_iter=2000,
    C=best_c,
    multi_class="multinomial"
)
clf.fit(X_tr, y_tr)

val_pred = clf.predict(X_val)
print("Validation macro-F1:", f1_score(y_val, val_pred, average="macro"))
print(classification_report(y_val, val_pred))
```

```python
# 6. Train FINAL model on ALL training data
final_clf = LogisticRegression(
    max_iter=2000,
    C=best_c,
    multi_class="multinomial"
)
final_clf.fit(X_train, y_train)

# 7. Predict test set + save submission
final_test_pred = final_clf.predict(X_test)

sub = pd.DataFrame({
    "id": test_ordered["id"],
    "emotion": final_test_pred
})

sub.to_csv("submission.csv", index=False)
print("Saved submission.csv")
```

## 1. Model Development (10 pts Required)

### 1.1 Preprocessing Steps

To start the project, I first combined the three datasets:
final_posts.json, data_identification.csv, and emotion.csv.
The JSON file had a nested structure, so I extracted only the fields we needed: the post_id (which I renamed to id) and the actual post text. After that, I merged the identification file to separate the training and test splits, and then added the emotion labels for the training IDs.
This resulted in a clean training dataframe with id, text, and emotion.

Once everything was merged, I focused on cleaning the text so the model could work with it more effectively. I converted all text to lowercase, replaced URLs with a URL token, and turned all user mentions into one consistent token: USER. Hashtags were simplified by removing the # symbol but keeping the word itself (for example, #angry becomes angry).
I also replaced numbers with a NUM token and removed punctuation and symbols that didn’t add emotional meaning, while keeping characters like ! and ? because they often reflect the intensity of a post.
Finally, I removed extra whitespace so every post had a clean, standardized format. These steps helped reduce noise and made the text much more suitable for feature extraction.

### 1.2 Feature Engineering Steps

After the preprocessing, I transformed the cleaned text into numerical features using TF–IDF. This method works well for short posts because it highlights which words (or combinations of words) are important in each message relative to the rest of the dataset.

I configured TF–IDF with a vocabulary limit of 40,000 features and used both unigrams and bigrams (ngram_range = (1, 2)). I also set min_df = 3 to remove extremely rare words that usually don’t help the model.
With these settings, I created:

* X_train: the TF–IDF feature matrix for all training posts

* X_test: the TF–IDF feature matrix for the posts in the test split

These matrices contain thousands of features but remain efficient to work with because they are stored as sparse vectors.

### 1.3 Explanation of Your Model

For the classification model, I used Logistic Regression, which is known to perform very well with high-dimensional text data like TF–IDF. I set max_iter = 2000 to make sure the model had enough iterations to fully converge, and I used the multinomial version since this is a multi-class emotion task.

Before training the final model, I created an 80/20 stratified train-validation split. Stratification ensures that all emotion categories appear proportionally in both sets, which makes the evaluation more reliable.
I evaluated the model using macro-averaged F1, because it gives equal weight to each emotion class and is the main metric used for this competition.

After checking the validation results and confirming that the model performed consistently, I retrained it on the full training dataset. I then used this final version to predict the emotions of the test posts and generate the submission.csv file in the correct format.
